Basic visualization using Matplotlib and Seaborn
Visualizations help you understand pollution levels, distributions, and relationships between variables.

A. Histogram
Shows how frequently different values occur.

B. Box plot
Shows the distribution, median, spread, and potential outliers.

C. Correlation heatmap
Shows how numerical variables relate to each other.

```python
print(df.columns.tolist())
```

```python
plt.plot(df["PM10"], marker='o')
```

```python
plt.title("PM10 LEVELS")
plt.xlabel("Record number")
plt.ylabel("PM10")

plt.show()
```

```python
pollutants = ['PM2.5', 'PM10', 'NOx', 'NO2', 'SO2', 'VOCs', 'CO', 'CO2', 'CH4', 'Temperature']
averages = df[pollutants].mean()
print(averages)
plt.bar(pollutants,averages)
plt.title("Average pollutant values")
plt.xlabel("Pollutants")
plt.ylabel("Averages")
```

```python
plt.hist(df["Temperature"].dropna(), bins=10, edgecolor="black")
#dropna() Pandas ka function hai jo missing values (NaN) ko remove karne ke liye use hota hai.
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(df["PM2.5"], df["PM10"])
#scatter() Matplotlib ka function hai jo do numerical variables ke beech relationship dikhane ke liye dots wala graph banata hai
plt.title("PM2.5 vs PM10")
plt.xlabel("PM2.5")
plt.ylabel("PM10")

plt.show()
```
