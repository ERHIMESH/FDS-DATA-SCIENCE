Q. A college want to analyze the marks obtained by 10 students in 5 subjects, write an python program using numpy and pandas to perform the following operation.

Q. create a marks matrix using default rng() function

```python
import numpy as np
import pandas as pd
```

```python
rng = np.random.default_rng()
names = ["Piyush","Himesh","Kavya","Avin","Ramsa"]
marks = rng.integers(0,101,size=(len(names),5))

df = pd.DataFrame(
    marks,
    index = names,
    columns = ["MATHS","SCIENCE","PHYSICS","COUMPTER","PYTHON"]
)
print(df)
```

Q. Calculate student perfomance find the total marks & percentage using numpy operation

```python
total_marks = np.sum(marks, axis=1)
for names, total in zip(names, total_marks):
    print(names, total)
```

```python
percentage = (total_marks/500)*100
df["Percentage"] = percentage
print(percentage)
for name, total, percent in zip(names, total_marks, percentage):
    print(name, total, f"{percent}%")
```

```python
result = pd.DataFrame({
    "Name": names,
    "Total": total_marks,
    "Percentage": percentage
})
print(result)
```

Q. Analyze the student perfomance i.e Calculate the student avg marks and std dev using numpy

```python
average_marks = np.average(total_marks)
print(average_marks)
```

```python
student_avg = np.mean(marks, axis=1)
student_std = np.std(marks, axis=1)

for name, avg , std in zip(names, student_avg, student_std):
    print(name, "Average: ",avg, "Standard dev: ", std)
```

Q. Identify pass and fail student using boolean marks ?

```python
passed = np.all(marks >= 40, axis=1)

for names, is_pass in zip(names, passed):
    print(names, "Pass" if is_pass else "Fail")
```
